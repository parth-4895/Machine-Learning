# Array Operations

```python
#vector-scaler arithmetic operations
print(a1+2)
print(a1-2)
print(a1*2)
print(a1/2)
print(a1**2)
```

```python
#vector-vector operations. This is possible here as the vectors have the same size
a2 = a1*9 - 33
print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
print(a1**a2)
```

```python
#relational
print(a1>5)
print(a2<0)
print(a1>a2)
```

# Array Functions Mathematical

```python
a1 = np.random.random((3,3))
a1= np.round(a1*100)
a1
```

```python
#max/min/sum/prod
# 0->column 1->row
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
print(np.max(a1, axis=1))#max of each row?
print(np.min(a1,axis = 1)) # min of each row?
print(np.sum(a1, axis = 0)) #sum of each column
print(np.prod(a1, axis = 0)) #prod of each column
```

```python
#mean/median/std/var
np.mean(a1, axis = 1) #same for all the other functions
```

```python
#trigonometric functions (mostly never used in data science career)
np.sin(a1)
```

```python
#dot product
a2 = np.arange(1,13).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

np.dot(a2,a3)
```

```python
#log and exponents
np.log(a1)
```

```python
np.exp(a1)
```

```python
#round/floor/ceil
a4 = np.random.random((2,3))*100
np.round(a4)
```

```python
np.ceil(a4)
```

```python
np.floor(a4)
```

# Indexing and Slicing

```python
a1 = np.arange(1,11)
a2 = np.arange(1,51).reshape(2,25)
a3 = np.arange(1,51).reshape(2,5,5)
```

```python
#indexing simple as it is normally. Fetching a number from an array using the index
```

```python
#slicing
a1[2:5] #all indexes [2,5)
```

```python
a1[2:8:2] # the last index gives the number of variables to leave
```

```python
a2
```

```python
a2[0:2, 3:6] #a2[rows, col]
```

```python
a4 = np.arange(0,12).reshape(3,4)
a4
```

```python
# now to fetch the values, 0,3,8,11
a4[::2, ::3] # the double colons ::x means that all rows/cols with x jumps
```

```python
a4[::2, 1:5:2]
```

```python
# OR
a4[::2 , 1::2]
```

```python
a3
```

```python
a3[0:2, 0:2, 0:2] # a3[the 2 dim arr, the rows of the one dim arr, the col of the one dim arr]
```

```python
a3[::2]
```
