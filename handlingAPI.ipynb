```python
df = pd.DataFrame();
```

```python
dfs=[]
```

```python
for i in range(1, 501): # Corrected range to fetch pages from 1 to 500 as per API limits
  response=requests.get('https://api.themoviedb.org/3/discover/movie?api_key=6b6fbb316656c1e0e65e07e7a1ebff2c&page={}'.format(i))
  if response.ok: # Check if the request was successful (status code 200)
    data = response.json()
    if 'results' in data: # Check if 'results' key exists
      temp_df = pd.DataFrame(data['results'])[['id', 'title', 'overview', 'release_date','popularity', 'vote_average', 'vote_count']]
      dfs.append(temp_df)
    else:
      print(f"Warning: 'results' key not found in response for page {i}. Response: {data}")
  else:
    print(f"Error: API request failed for page {i} with status code {response.status_code}. Response: {response.text}")

df = pd.concat(dfs, ignore_index=True)
```
